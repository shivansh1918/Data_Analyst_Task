```python
df["Bonus"] = df["Bonus"] + 2000 #update your data
```

```python
df["salary"] =df["salary"].fillna("unknown values")   #for filling with constant
```

```python
df["Income"] = df["Income"].fillna()(df["Income"].mean())    #numerical and data should not contain any outliers
```

```python
df["salary"] = df["salary"].fillna()(df["salary"].median())  #numerical and data should contain  outliers
```

```python
df["salary"] = df["salary"].fillna()(df["salary"].mode())    #categorical and text kind of data
```

```python
df.duplicated()
```

```python
df.drop_duplicates(
    subset = ["Income"]
    inplace= True
)
```
